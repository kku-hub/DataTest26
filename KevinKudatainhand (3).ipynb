## Project Question

Which college football has the most all time wins?

My Project will be comparing all College Football D1 FBS and FCS team to find who has the most win in history.

## Data Collection

All of my data source come from Sports Reference College Football. This includes team name, total wins and total lost.

Source: https://www.sports-reference.com/cfb/schools/

## Data Structure

My dataset has three columns: Team, Wins, ans Lost each row will represent a College Football team. 

```python
import pandas as pd


df= pd.read_csv("college_football.csv",header=[0,1])

df.head()
```

```python
df["Overall","W"]=pd.to_numeric(
    
df["Overall","W"],errors="coerce"
    
)

top_teams=df.sort_values(
   by= ("Overall","W"), 
    ascending=False
).reset_index(drop=True)

top_teams.index=top_teams.index+1

top_teams.head(10)


top_teams[[("Unnamed: 1_level_0","School"), ("Overall","W"),("Overall","L")]].head(10)
```
