## Analysis Plan

My Analysis plan is to find out college football team base on all time wins. I will sort teams from the most wins to the least wins. I will also compare their losses and create a bar chart to see the top 10 teams. 

## Justifcation 

I am going to choose to sort teams by wins because it help me decide which college football teams has the best winning history. I will use bar chart for this assignment.

```python
import pandas as pd

df=pd.read_csv("college_football.csv",header=1)
df.head()
```

```python
print(df.columns.tolist())
```

```python
top10 = df.sort_values("W", ascending=False).head(10)

top10
```

```python
import matplotlib.pyplot as plt

top10.plot(x="School", y="W", kind="bar")

plt.title("Top 10 College football Team wins by Wins")

plt.xlabel("College Football Teams")


plt.ylabel("Total wins")

plt.show()




```
